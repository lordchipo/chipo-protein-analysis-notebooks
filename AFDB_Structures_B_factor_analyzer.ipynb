# B-factor analyzer for AlphaFold Protein Structure Database protein structures

## 1. Install and import the required libraries/packages
- Biopandas for handling the protein structure file as a DataFrame
- Matplotlib for plotting

```python
!pip install biopandas
!pip install matplotlib
from biopandas.mmcif import PandasMmcif
import matplotlib.pyplot as plt
from matplotlib import style
```

## 2. Enter UniProt ID of protein to fetch its structure from AlphaFold DB

```python
prot_id = input("Enter the protein PDB ID or UniProt (AlphaFold DB) ID: ")
structure = PandasMmcif().fetch_mmcif(uniprot_id = prot_id, source = 'alphafold2-v4')

# Display raw contents of the protein's mmCIF file from AlphaFold DB
print("\nRaw mmCIF file contents:\n\n%s\n..." % structure.mmcif_text[:2000])
```

## 3. Preview of the protein structure DataFrame
Display first 20 rows of the DataFrame

```python
structure.df['ATOM'][:20]
```

## 4. Calculate average B-factors for the protein backbone

```python
backbone = structure.df['ATOM'][(structure.df['ATOM']['label_atom_id'] == 'C') |
                               (structure.df['ATOM']['label_atom_id'] == 'CA')|
                               (structure.df['ATOM']['label_atom_id'] == 'N') |
                               (structure.df['ATOM']['label_atom_id'] == 'O')]
average_b_factors = backbone['B_iso_or_equiv'].mean()
print(f'Average B-Factor: {average_b_factors} ')
```

## 5. Display distribution of B-factors

```python
style.use('ggplot')
structure.df['ATOM']['B_iso_or_equiv'].plot(kind='hist')
plt.title('Distribution of B-factors')
plt.xlabel('B-Factor')
plt.ylabel('Frequency')
plt.show()
```

## 6. Display B-factors along the polypeptide chain

```python
structure.df['ATOM']['B_iso_or_equiv'].plot(kind='line')
plt.title('B-Factors Along the Amino Acid Chain')
plt.xlabel('Atom Number')
plt.ylabel('B-factor ($A^2$)')
plt.show()
```

### Interpretation of results:
- The B-factor is a measure of an atom's displacement from a mean position in a protein crystal.
- Higher B factors are associated with regions that are more flexible and dynamic, while lower B factors indicate more ordered and rigid regions.
- B-factor values are normally between 15 to 30 A^2, but are often much higher for flexible regions.
