# Protein Domain Finder from Sequence

## 1. Install biopython and import the required modules
- Install Biopython into the Google Colab runtime
- Import Prosite and ScanProsite from Bio.ExPASy for domain search and description
- Import the SeqIO module to enable input of sequences
- Import "files" from the google.colab library

```python
!pip install biopython
from Bio import SeqIO
from Bio import ExPASy
from Bio.ExPASy import Prosite
from Bio.ExPASy import ScanProsite
from google.colab import files
```

## 2. Enter a protein query sequence file
The file format for the query is *.fasta

```python
file_upload = files.upload()
query_dir = list(file_upload.keys())[0]

query = SeqIO.read(query_dir, format = "fasta")
print(query.description)
print("Length of the query protein is: " + str(len(query)))
print(query.seq)
```

## 3. Find domain in given protein sequence using ScanProsite


```python
handle = ScanProsite.scan(seq=query.seq, mirror="https://prosite.expasy.org/") # generates a handle variable to scan a given sequence and determine the domain
result = ScanProsite.read(handle)                                              # reads the result of the domain search on the given protein sequence
result.n_match 				                                                         # returns the number of matches between the given sequence and the Prosite domain
                                                                               # from the Prosite domain search
if result.n_match == 0:
  print("No match found")
elif result.n_match == 1:
  print("1 match found")
else:
  print(str(result.n_match) + " matches found")

x = 0
while x < result.n_match:
  print(result[x])
  x += 1
```

## 4. Describe each domain and pattern found on protein sequence using Prosite

```python
y = 0
while y < result.n_match:
  domain_handle = ExPASy.get_prosite_raw(result[y]['signature_ac'])
  domain_record = Prosite.read(domain_handle)
  print(result[y]['signature_ac'])
  print("Start of domain: " + str(result[y]['start']) + ", End of domain: " + str(result[y]['stop']))
  print(domain_record.description)
  print(domain_record.pattern)
  print(query.seq[result[y]['start']:result[y]['stop']])
  print("\n")
  y += 1
```
